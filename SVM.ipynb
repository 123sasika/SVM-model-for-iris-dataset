```python
# creating a dataframe via pandas
# displaying data and feature names (columns)
df = pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
```

```python
# artificially we are introducing a new column to the dataset
df['target'] = iris.target
df.head()
```

```python
# introducing new field as flower name.
# lambda function is used to connect target field and target names field.
# so we can display the flower name as well in the same dataset
df['flower_name'] =df.target.apply(lambda x: iris.target_names[x])
df
```

```python
# segmenting the dataset into three categories and introduce 3 different data frame variables to store
# them seperately.
df0 = df[:50]
df1 = df[50:100]
df2 = df[100:]
```

```python
from sklearn.model_selection import train_test_split
```

```python
iv = df.drop(['target','flower_name'], axis='columns')
dv = df.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(iv, dv, test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
X_test
```

```python
from sklearn.svm import SVC
model = SVC()
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
model_k = SVC(kernel="rbf")
model_k.fit(X_train, y_train)
model_k.score(X_test, y_test)
```

```python
model_C = SVC(C=10)
model_C.fit(X_train, y_train)
model_C.score(X_test, y_test)
```

```python
model_C = SVC(C=0.1)
model_C.fit(X_train, y_train)
model_C.score(X_test, y_test)
```

```python
model_g = SVC(gamma=1)
model_g.fit(X_train, y_train)
model_g.score(X_test, y_test)
```

```python
param_grid = {'C':[0.1,1,10], 'gamma':[0.01,0.1,1]}

from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(SVC(), param_grid)
grid.fit(X_train, y_train)
grid.score(X_test, y_test)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
# get the predictions for the X_test
y_predicted=grid.predict(X_test)
```

```python
y_predicted
```

```python
# a confusion metrics is created to represent correct and wrong predictions.
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_predicted)
```
